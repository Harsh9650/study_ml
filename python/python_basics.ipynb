## **Indexing**: Syntax is given as, example[Start : Stop : Stepover]

- **Start default value is 0**,
- **Stop default value is the length of the sequence (means the last element)**,
- **Stepover default value is 1 (one element per step)**
- **So, example[0] is same as example[0:1] or example[0:1:1]**

```python
# A string is an ordered collection of characters. 

example_string = '0123456789'

"""In this string, we have 10 characters. The first element of the sequence will be at index 0.
    So, 5th element of the sequence can be accessed using index 4 and so on."""

example_string[0], example_string[0:1], example_string[0:1:1]
```

### The last element of the sequence can be accessed using -1 or the (length of the sequence - 1).

```python
string_len = len(example_string) # built-in function to get the length of an iterable.

example_string[string_len - 1] == example_string[-1] 
```

### [start:stop], the stopping index doesn't include itself, so the last element will be at index 'stop - 1'.

```python
example_string[0:4], example_string[3]  
```

## **Sorting a list**

> ### `.sort()` is a method that is used to sort the elements of a list.
>
>> Deafult sorting is in ascending order (smallest first).
>> 
>> `reverse` is parameter that is used to arrange the elements in deascending order.
>>
>>  Note:- just like .append(), .insert(), and others .sort() doesn't return anything (returns None), hence it rearranges elements on the fly.
>
> ### `Sorted()`is a function that can be used to sort the elements of a list.
> 
>> Unlike .sort() method it returns a new list of the sorted elements.
>>
>> Again deafult sorting is in ascending order (smallest first).
>>
>> The parameter `reverse` is used to arrange the elements in deascending order.
>>
>> **Note:- calling a sorted function on a list doesn't amend that list, it creates a new list and expect you to save it in a variable.**

```python
list_sample1, list_sample2 = ['1', '2', '1', '3', '2', '4', '1', '3'], ['a', 'b', 'a', 'c', 'b', 'd', 'a', 'c']

list_sample1.sort()
list_sample2.sort()

list_sample1, list_sample2 
```

```python
list_sample1, list_sample2 = ['1', '2', '1', '3', '2', '4', '1', '3'], ['a', 'b', 'a', 'c', 'b', 'd', 'a', 'c']

test3 = sorted(list_sample1)
test4 = sorted(list_sample2)
test5 = sorted(list_sample1, reverse = True)
test6 = sorted(list_sample2, reverse = True)

list_sample1, list_sample2, test3, test4, test5, test6 # old lists still have the same order.
```

## **Methods for Sets:-**

---

- #### `.difference()` returns the number of elements in set1 that are not present in set2.
- #### `.discard()` is used to remove an element from an exiting set, and it returns None.
- #### `.difference_update()` is used to remove the elements from set1 that are present in the set2, it returns None.
- #### `.intersection()` returns the common elements between two sets.
- #### `.isdisjoint()` returns true if both sets have nothing in common or false if they have some common elements.
- #### `.issubset()` returns True, if all elements in set1 are present in set2, otherwise False. This means True when set2 contains complete set1.
- #### `.issuperset()` It returns True, if all elements in set2 are present in set1, otherwise False. This means True when set1 contains complete set2.
- #### `.union()` It returns a new set by combining the elements from the two sets. Again no duplicates, so repeating elements are discarded.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8, 9, 10}

m1 = set1.difference(set2) # This returns the unique elements in set1 that are not in set2.
m1
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8, 9, 10}

set1.discard(3)
set1
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8, 9, 10}

set1.difference_update(set2) 
set1
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8, 9, 10}

m2 = set1.intersection(set2) 
m2
```

```python
# another way to perform intersection between two sets, is to use the & operator.

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8, 9, 10}

m3 = set1 & set2 
m3
```

## **Note:- When we create a function and use a variable or a different function in it, the python interpreter will**

- ####  First check the existence of that variable/function inside it's local (function) environment.
- ####  If the used variable/function doesn't exist in the local (function) environment then it will search for that variable in the parent scope.
- ####  In case of nested functions, parent scope refers to the first function that contains the second function.
- ####  In case of single functions, parent scope refers to global environment.
- ####  In case we are calling a function within a function, and that function doesn't exist in the entire python file (local and global scope), then python interpreter will look for that function in the list of pythons built-in function. if it exists it use, if it didn't exist, it throws an error.

```python
a = 1 

def parent_function():
    a = 10
    def confusion():
        return a
    return confusion()

print(a) # global variable
print(parent_function())
```

## **nonlocal** keyword is used to tell the python interpreter that we want to use the a variable -- but from our local environment.

```python
# without nonlocal

def outer():
    x = "local"
    def inner(): 
        x = "nonlocal"
        print("inner", x)
    inner()
    print("outer", x)

outer() # different values of x for global and local environment.
```

```python
def outer():
    """ This function is created to demonstrate the use of nonlocal keyword. Here, by using nonlocal
        keyword we are telling the python interpreter that we want to use the variable x but 
        from our local environment, and you don't need to create a new variable x.
        Important to understand is that parent must not be global.
    """
    x = "local"
    def inner():
        nonlocal x
        x = "nonlocal" # now, we re-assign the x variable from our parent scope with "nonlocal" value.
        print("inner", x)
    inner()
    print("outer", x)

outer()
```

```python
x = "local"

def outer():
    """ This function is created to demonstrate the difference between global and nonlocal keyword. Here, by using global
        keyword we are telling the python interpreter that we want to use the keyword x but 
        from our global environment. 
    """
    def inner():
        global x # Since, x is not defined in the global environment.
        x = "nonlocal" # now, we re-assign the global x variable with "nonlocal" value.
        print("inner", x)
    inner()
    print("outer", x)

outer()
```
